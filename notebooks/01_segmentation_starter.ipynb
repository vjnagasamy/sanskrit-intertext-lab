# Sanskrit Segmentation Starter

Demonstrates normalization and sentence segmentation using the `DandasSegmenter` (daṇḍa-based regex).

**Source formats supported:** `devanagari` (default) or `iast` (IAST romanization → Devanagari).

```python
import sys
from pathlib import Path

# Add project root to sys.path so imports work from notebooks/
PROJECT_ROOT = Path.cwd().parent
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from sanskrit_pipeline.sdk import SanskritResearchSDK
```

```python
sdk = SanskritResearchSDK(
    engine="dandas",
    source_format="devanagari",
    split_on_single_danda=False,  # set True for pāda-level segments
)
```

```python
# Sample Sanskrit text — Bhagavad Gita 2.47-2.48
sample_text = """
कर्मण्येवाधिकारस्ते मा फलेषु कदाचन।
मा कर्मफलहेतुर्भूर्मा ते सङ्गोऽस्त्वकर्मणि॥
योगस्थः कुरु कर्माणि सङ्गं त्यक्त्वा धनंजय।
सिद्ध्यसिद्ध्योः समो भूत्वा समत्वं योग उच्यते॥
"""

seg_view = sdk.segment_text(sample_text)
print(f"Engine: {seg_view.engine_name}")
print(f"Segments: {len(seg_view.segments)}")
print()
for segment in seg_view.segments:
    print(segment)
```

```python
seg_view.to_dataframe()
```

```python
# Pāda-level segmentation: also split on single daṇḍa (।)
sdk_pada = SanskritResearchSDK(
    engine="dandas",
    source_format="devanagari",
    split_on_single_danda=True,
)

seg_pada = sdk_pada.segment_text(sample_text)
print(f"Pāda-level segments: {len(seg_pada.segments)}")
for segment in seg_pada.segments:
    print(segment)
```

```python
# IAST romanization input — automatically converted to Devanagari
# Sample Sanskrit text — Manusmriti (8.15), Mundaka Upanishad (3.1.6)
iast_text = "dharmo rakṣati rakṣitaḥ॥ satyameva jayate॥"

sdk_iast = SanskritResearchSDK(engine="dandas", source_format="iast")
seg_iast = sdk_iast.segment_text(iast_text)
print(f"Normalized (Devanagari): {seg_iast.normalized_text}")
for segment in seg_iast.segments:
    print(segment)
```
